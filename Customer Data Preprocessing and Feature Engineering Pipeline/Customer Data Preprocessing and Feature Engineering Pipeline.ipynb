# Task 3 – Data Cleaning

```python
#1. Missing Value Imputation

# Missing values before cleaning
missing_before = df.isnull().sum()
missing_before[missing_before > 0]
```

```python
# Creating a copy for cleaning
df_clean = df.copy()

# Numerical columns -> Using Median Imputation
num_cols = df_clean.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Categorical columns -> Using Mode Imputation
cat_cols = df_clean.select_dtypes(include='object').columns

for col in cat_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
```

```python
#AFTER
missing_after = df_clean.isnull().sum()
missing_after[missing_after > 0]
```


**Reason:** Missing values were imputed instead of removing records to preserve valuable information. **Median imputation** was used for numerical attributes because it is less affected by outliers, while **Mode imputation** was used for categorical attributes to replace missing values with the most frequently occurring category.

# Task 6 – Proximity Measures

```python
#Step 1: Select 20 Records
sample_df = reduced_df.sample(n=20, random_state=42)
sample_df.reset_index(drop=True, inplace=True)

sample_df.head()
```

```python
#Step 2: Select Numerical Features

numeric_sample = sample_df.select_dtypes(include=['int64','float64'])

numeric_sample.head()
```

```python
#Measure 1 – Euclidean Distance
from sklearn.metrics.pairwise import euclidean_distances

euclidean_matrix = euclidean_distances(numeric_sample)

euclidean_df = pd.DataFrame(
    euclidean_matrix,
    index=sample_df.index,
    columns=sample_df.index
)

euclidean_df
```

**INTERPRETAION:** Euclidean Distance measures the straight-line distance between two customers based on their numerical features. Smaller distance values indicate that two customers have more similar characteristics, while larger values indicate greater differences.

```python
#Measure 2 – Cosine Similarity

from sklearn.metrics.pairwise import cosine_similarity

cosine_matrix = cosine_similarity(numeric_sample)

cosine_df = pd.DataFrame(
    cosine_matrix,
    index=sample_df.index,
    columns=sample_df.index
)

cosine_df
```

**INTERPRETATION:** Cosine Similarity measures the similarity between customers based on the direction of their feature vectors rather than their absolute values. Values closer to 1 indicate highly similar customers, while values closer to 0 indicate low similarity.

```python
#Identify the Most Similar Customers

#Euclidean Distance

import numpy as np

euclidean_copy = euclidean_df.copy()

np.fill_diagonal(euclidean_copy.values, np.inf)

closest_pair = np.unravel_index(
    np.argmin(euclidean_copy.values),
    euclidean_copy.shape
)

print("Most Similar Customers (Euclidean):")
print("Customer", closest_pair[0], "and Customer", closest_pair[1])

print("Distance:",
      euclidean_copy.iloc[closest_pair])
```

```python
#Cosine Similarity

cosine_copy = cosine_df.copy()

np.fill_diagonal(cosine_copy.values, -1)

most_similar = np.unravel_index(
    np.argmax(cosine_copy.values),
    cosine_copy.shape
)

print("Most Similar Customers (Cosine):")
print("Customer", most_similar[0], "and Customer", most_similar[1])

print("Similarity:",
      cosine_copy.iloc[most_similar])
```

| Proximity Measure  | Purpose                                                               | Interpretation                                                |
| ------------------ | --------------------------------------------------------------------- | ------------------------------------------------------------- |
| Euclidean Distance | Measures the straight-line distance between customer feature vectors. | Smaller values indicate greater similarity between customers. |
| Cosine Similarity  | Measures the similarity in the direction of customer feature vectors. | Values closer to 1 indicate highly similar customers.         |



**Proximity Measures**

A random sample of 20 customer records was selected from the cleaned dataset to analyze customer similarity. Two proximity measures were applied: Euclidean Distance and Cosine Similarity. Euclidean Distance was used to measure the direct numerical distance between customers, where smaller values indicate greater similarity. Cosine Similarity was used to compare the orientation of customer feature vectors, with values closer to 1 representing highly similar customer profiles. The pair of customers with the smallest Euclidean distance and the highest Cosine Similarity were identified as the most similar customers in the selected sample.
